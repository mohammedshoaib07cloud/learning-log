```python
import torch
from torch.utils.data import DataLoader
from torchvision import datasets
from torchvision.transforms import ToTensor
import matplotlib.pyplot as plt
```

```python
#loading the dataset
data=datasets.FashionMNIST(root="data",train=True,download=True,transform=ToTensor())
print(len(data))
img, label=data[0]
print(img.shape,label)
```

```python
#looking at the sample of the data
labels_map={0: "T-Shirt", 1: "Trouser", 2: "Pullover", 3: "Dress", 4: "Coat",
              5: "Sandal", 6: "Shirt", 7: "Sneaker", 8: "Bag", 9: "Ankle Boot"}
plt.figure(figsize=(8, 8))
for i in range(1,10):
    idx = torch.randint(len(data), size=(1,)).item()
    img, lanel= data[idx]
    plt.subplot(3,3,i)
    plt.title(labels_map[label])
    plt.imshow(img.squeeze(),cmap="gray")
plt.show()
```

```python
#Dataloaders=how the data will be loaded for training
#batch shape,size=In how much size will the data loaded for the training process
loader= DataLoader(data,batch_size=64,shuffle=True,num_workers=0)
X, y =next(iter(loader))
print(X.shape,y.shape)
```
